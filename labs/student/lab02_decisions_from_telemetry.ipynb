## 1. Comparisons produce Boolean values

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `<=` | less than or equal to |
| `>` | greater than |
| `>=` | greater than or equal to |

## 4. Combining conditions

- `A and B` is true only when both conditions are true.
- `A or B` is true when at least one condition is true.
- `not A` reverses a Boolean value.

### PREDICT 7

````python
battery_charge = 30
bus_voltage_v = 27
comms_ok = False

print(battery_charge >= 25 and bus_voltage_v >= 24)
print(battery_charge < 25 or bus_voltage_v < 24)
print(comms_ok)
print(not comms_ok)
````
